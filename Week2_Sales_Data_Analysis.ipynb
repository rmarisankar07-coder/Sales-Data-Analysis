```python
import pandas as pd
import random
from datetime import datetime, timedelta

# DATASET CREATION FUNCTION
def generate_sales_data(num_records=60):
    catalog = {
        "ELECTRONICS": [
            ('Smartphone', 600, 900),
            ('TWS', 40, 100),
            ('Smart Watch', 150, 300)
        ],
        "CLOTHING": [
            ('Denim Jeans', 40, 80),
            ('Cotton Shirt', 15, 50),
            ('Formal Pant', 60, 100),
            ('Hoodie', 75, 90)
        ],
        'KITCHEN ELECTRONICS': [
            ('Blender', 30, 50),
            ('Coffee Maker', 45, 75),
            ('Electric Stove', 50, 100)
        ],
        'CONSUMER ELECTRONICS': [
            ('LED Light', 30, 75),
            ('Electric Fan', 60, 80),
            ('Refrigerator', 50, 200)
        ]
    }

    data = []
    end_date = datetime.now()
    start_date = end_date - timedelta(days=100)

    for i in range(num_records):
        # Order ID
        order_id = i + 1

        # To generate Random date
        random_days = random.randint(0,100)
        order_date = start_date + timedelta(days=random_days)

        # To select Category and Products
        category = random.choice(list(catalog.keys()))
        product_info = random.choice(catalog[category])
        product_name = product_info[0]

        # To generate Price and Quantity
        unit_price = round(random.uniform(product_info[1], product_info[2]), 2)
        quantity = random.randint(1,8)

        # To calculate Total Amount
        total_amount = round(quantity*unit_price, 2)

        data.append([
            order_id,
            order_date.strftime('%Y-%m-%d'),
            product_name,
            category,
            quantity,
            unit_price,
            total_amount
        ])

    columns=['ORDER_ID', 'ORDER_DATE', 'PRODUCT_NAME', 'CATEGORY', 'QUANTITY', 'UNIT_PRICE', 'TOTAL_AMOUNT']
    return pd.DataFrame(data, columns=columns)

# Generate dataset
df = generate_sales_data(60)

# Analysis
df['ORDER_DATE'] = pd.to_datetime(df['ORDER_DATE'])
total_revenue = df['TOTAL_AMOUNT'].sum()
total_orders = df.shape[0]
avg_order_value = total_revenue / total_orders

# Product & Category Performance
category_analysis = df.groupby('CATEGORY')['TOTAL_AMOUNT'].sum().sort_values(ascending=False)
product_analysis = df.groupby('PRODUCT_NAME')['TOTAL_AMOUNT'].sum().sort_values(ascending=False)

# Monthly Trend
df['MONTH'] = df['ORDER_DATE'].dt.to_period('M')
monthly_analysis = df.groupby('MONTH')['TOTAL_AMOUNT'].sum()

# BUSINESS SUMMARY REPORT
print("--- BUSINESS SUMMARY ---")
print(f"1. Overall Performance: The store generated a total revenue of ${total_revenue:,.2f} across {total_orders} orders.")
print(f"2. Top Category: '{category_analysis.index[0]}' is the highest performing category.")
print(f"3. Best Seller: The '{product_analysis.index[0]}' contributed the most to sales revenue.")
print(f"4. Monthly Insight: Analysis covers the last 6 months to identify seasonal trends.")
print("\n--- DATA PREVIEW (First 5 Rows) ---")
print(df.drop(columns=['MONTH']).head())

# EXPORT DATASET
output_filename = 'week2_sales_analysis_output.csv'
final_df = df.drop(columns=['MONTH'])
final_df.to_csv(output_filename, index=False)

```
